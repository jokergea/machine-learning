```python
import pandas as pd
import numpy as np
import re
from sklearn.model_selection import train_test_split, cross_val_score, KFold
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LinearRegression

df = pd.read_csv("AI_Usage_and_Impact_on_Students_and_Professionals.csv")

def clean_numeric(val):
    if pd.isna(val): return np.nan
    nums = re.findall(r"[-+]?\d*\.\d+|\d+", str(val))
    return float(nums[0]) if nums else np.nan

numeric_cols = ['Age', 'Monthly_Income', 'AI_Usage_Hours_Per_Day',
                'Monthly_AI_Cost', 'Time_Saved_Hours_Per_Week', 'Work_or_Study_Hours_Per_Day']
for col in numeric_cols:
    df[col] = df[col].apply(clean_numeric)

for col in numeric_cols + ['Productivity_Score', 'Satisfaction_Score', 'Accuracy_Rating']:
    df[col].fillna(df[col].median(), inplace=True)

categorical_cols = ['Gender', 'Country', 'User_Type', 'Education_Level', 'AI_Tool', 'Would_Recommend']
for col in categorical_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

encoder = LabelEncoder()
for col in categorical_cols:
    df[col] = encoder.fit_transform(df[col].astype(str))

X = df[numeric_cols + categorical_cols].drop(columns=['Would_Recommend'])
y = df['Productivity_Score']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

lin_model = LinearRegression()
lin_model.fit(X_train, y_train)
y_pred_lin = lin_model.predict(X_test)

print(f"Intercept: {lin_model.intercept_:.3f}")
print(f"First 3 Coefficients: {lin_model.coef_[:3]}")

kf = KFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(lin_model, X_scaled, y, cv=kf, scoring='r2')

print(f"R2 CV Scores: {cv_scores}")
print(f"Average CV R2: {cv_scores.mean():.3f}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score

y_clf = df['Would_Recommend']
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_scaled, y_clf, test_size=0.2, random_state=42)

log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train_c, y_train_c)
y_pred_log = log_model.predict(X_test_c)

print(f"Accuracy:  {accuracy_score(y_test_c, y_pred_log):.3f}")
print(f"Precision: {precision_score(y_test_c, y_pred_log, average='weighted', zero_division=0):.3f}")
print(f"Recall:    {recall_score(y_test_c, y_pred_log, average='weighted', zero_division=0):.3f}")
print(f"F1-Score:  {f1_score(y_test_c, y_pred_log, average='weighted', zero_division=0):.3f}")
print("Confusion Matrix:\n", confusion_matrix(y_test_c, y_pred_log))

nb_model = GaussianNB()
nb_model.fit(X_train_c, y_train_c)
y_pred_nb = nb_model.predict(X_test_c)

print(f"Gaussian NB Accuracy: {accuracy_score(y_test_c, y_pred_nb):.3f}")

svm_model = SVC(kernel='rbf', C=1.0, random_state=42)
svm_model.fit(X_train_c, y_train_c)
y_pred_svm = svm_model.predict(X_test_c)

print(f"SVM Accuracy: {accuracy_score(y_test_c, y_pred_svm):.3f}")
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print(f"Original Feature Count: {X_scaled.shape[1]}")
print(f"Reduced Feature Count: {X_pca.shape[1]}")
print(f"Variance captured by first 2 components: {sum(pca.explained_variance_ratio_) * 100:.2f}%")
```
